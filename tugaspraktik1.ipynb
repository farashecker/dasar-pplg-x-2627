```python
%%writefile penjumlahan.cpp
#include <iostream>
using namespace std;

int main(){
    int a,b,c,d;

    cout << "masukkan nilai a :";
    cin >> a;
    cout << "masukkan nilai b :";
    cin >> b;

    c = a+b;
    d = a-b;

    cout << "hasil penjumlahan: " << c << endl;
    cout << "hasil pengurangan: " << d << endl;

    return 0;
}
```
